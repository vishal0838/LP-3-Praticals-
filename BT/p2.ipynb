## Step 3: Understand the wallet

Your MetaMask wallet provides the following features:

- Account address: Used to receive cryptocurrency.
- Send: Transfer supported crypto assets to another address.
- Receive: Display or copy your public address.
- Activity: View your transaction history.
- Network selector: Switch between supported blockchain networks.

## Step 4: Test a transaction safely

For a college practical, use the Ethereum Sepolia test network rather than real money.

1. Select Sepolia in MetaMask, enabling test networks in settings if necessary.
2. Obtain test ETH from a reputable Sepolia faucet.
3. Use a second test account as the recipient.
4. Select Send, enter the recipient's public address, and enter a small test amount.
5. Review the transaction fee and confirm.
6. Open the transaction details to inspect its status and fee.

Test ETH has no real monetary value. Never use your real recovery phrase on a faucet or share your private key.

# Practical: Installation of MetaMask and Study of Ether Spent per Transaction

Aim: To install MetaMask, create an Ethereum wallet, perform a test transaction, and study the amount of Ether (ETH) spent per transaction.

## Step 2: Create an Ethereum wallet

1. Open the MetaMask extension.
2. Click Create a new wallet.
3. Follow the setup instructions and create a strong password.
4. Securely back up your Secret Recovery Phrase offline.
5. Open your wallet and locate your public account address.

Security: Never share your Secret Recovery Phrase or private key with anyone. Do not use real funds for this college practical.

## Step 3: Select a test network

For a safe demonstration, use the Sepolia test network.

1. Open MetaMask.
2. Open the network selector.
3. Enable or select Sepolia testnet if available.
4. Obtain test ETH from a reputable Sepolia faucet, if you need funds for the experiment.

Test ETH has no real monetary value. Ethereum transactions on the main network use real ETH for transaction fees.&#x20;

## Step 4: Perform a test transaction

You can use two MetaMask accounts that you control.

1. Create a second account in MetaMask, or use another test account.
2. Copy the recipient's public address.
3. Select Send in your first account.
4. Paste the recipient address.
5. Enter a small amount of Sepolia test ETH.
6. Review the estimated network fee.
7. Click Confirm.
8. Wait for the transaction to be confirmed.

Do not send real ETH to complete this practical.

## Step 5: Study Ether spent per transaction

Open the transaction details in MetaMask. Record the following information.

| Parameter           | Meaning                       |
| ------------------- | ----------------------------- |
| Transaction hash    | Unique transaction identifier |
| Amount sent         | ETH transferred to recipient  |
| Gas used            | Computational units consumed  |
| Effective gas price | Price paid per gas unit       |
| Transaction fee     | ETH paid for the transaction  |

The basic calculation is:

\\[ \text{Transaction Fee (ETH)} =\frac{\text{Gas Used}\times\text{Gas Price (Gwei)}}{10^9} \\]

For example, suppose a transaction uses 21,000 gas at 10 Gwei:

\\[ \frac{21,000\times10}{10^9} =0.00021\text{ ETH} \\]

This is an illustrative example, not a live fee estimate. The actual fee depends on network conditions and transaction type.

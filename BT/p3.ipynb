## 4. Understanding the code

- `mapping(address => uint256)`: stores a separate simulated balance for each wallet address.
- `deposit()`: accepts Ether and credits the caller's balance.
- `withdraw()`: checks the caller's balance, deducts the requested amount, and transfers the Ether.
- `getBalance()`: returns the balance belonging to the caller.
- `require()`: prevents invalid operations, such as withdrawing more than the available balance.

Write a program in solidity to create Student data

Yes! You can demonstrate your Decentralized E-Voting System without deploying a smart contract to any blockchain. The easiest way is to run a local demo in your browser using HTML, CSS, and JavaScript.

This demo lets you connect no wallet, cast votes for candidates, prevent repeat voting in the current browser session, and display results. It is a UI simulation, not a real blockchain dApp.

## Step 1: Create the demo

1. Open VS Code.
2. Create a folder named `Voting-Demo`.
3. Inside it, create a file named `index.html`.
4. Copy the following code into the file.

```

<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>E-Voting Demo</title>

<style>
body {
    font-family: Arial, sans-serif;
    background: #f0f4f8;
    text-align: center;
    padding: 25px;
}
h1 { color: #173b70; }
#notice { color: #555; }
.card {
    background: white;
    padding: 20px;
    margin: 15px auto;
    max-width: 350px;
    border-radius: 12px;
    box-shadow: 0 3px 10px #ccc;
}
button {
    padding: 10px 18px;
    background: #2059a6;
    color: white;
    border: none;
    border-radius: 6px;
    cursor: pointer;
}
button:disabled { background: gray; }
</style>
</head>

<body>
<h1>🗳️ E-Voting System</h1>
<p id="notice">Demo mode — no blockchain deployment required.</p>
<p id="status">Choose a candidate to cast your demo vote.</p>

<div id="candidates"></div>
<button onclick="resetDemo()">Reset Demo</button>

<script>
const candidates = [
    { id: 1, name: "Candidate A", votes: 0 },
    { id: 2, name: "Candidate B", votes: 0 },
    { id: 3, name: "Candidate C", votes: 0 }
];

let voted = false;

function showCandidates() {
    const area = document.getElementById("candidates");
    area.innerHTML = "";

    candidates.forEach(candidate => {
        const card = document.createElement("div");
        card.className = "card";

        card.innerHTML = `
            <h2>${candidate.name}</h2>
            <p>Votes: ${candidate.votes}</p>
            <button ${voted ? "disabled" : ""}
                onclick="vote(${candidate.id})">
                ${voted ? "Vote Submitted" : "Vote"}
            </button>
        `;

        area.appendChild(card);
    });
}

function vote(id) {
    if (voted) return;

    const candidate = candidates.find(c => c.id === id);
    candidate.votes++;
    voted = true;

    document.getElementById("status").innerText =
        "Demo vote recorded for " + candidate.name + "!";

    showCandidates();
}

function resetDemo() {
    candidates.forEach(c => c.votes = 0);
    voted = false;

    document.getElementById("status").innerText =
        "Demo reset. You can vote again.";

    showCandidates();
}

showCandidates();
</script>
</body>
</html>

```

## Step 2: Run the demo

Save the file

Press `Ctrl + S` in VS Code.

Open in your browser

Find `index.html` in File Explorer and double-click it. Chrome or Edge will open the demo.

Demonstrate voting

Click a candidate's Vote button. The vote count increases, and the buttons become disabled to prevent another vote in that demo.

## Step 3: What to show your teacher

- The three candidates and their initial vote counts.
- Casting a vote and seeing the count update.
- Attempting to vote again.
- Using Reset Demo to start over.

Note: This version does not use Solidity, MetaMask, or a blockchain. The votes exist only in browser memory and disappear when the page is refreshed. Describe it as a frontend simulation of an e-voting dApp, not a deployed decentralized application.

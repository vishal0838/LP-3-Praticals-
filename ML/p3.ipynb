Implement Gradient Descent Algorithm to find the local minima of a function. For example, find the local minima of the function y=(x+3)² starting from the point x=2

```python

# Gradient Descent Algorithm

# Define the function
def function(x):
    return (x + 3) ** 2

# Derivative of the function
def gradient(x):
    return 2 * (x + 3)

# Initial point
x = 2

# Learning rate
learning_rate = 0.1

# Number of iterations
iterations = 30

print("Iteration\t x\t\t y")

for i in range(iterations):
    y = function(x)
    print(i, "\t\t", round(x, 4), "\t\t", round(y, 4))

    # Update x using Gradient Descent
    x = x - learning_rate * gradient(x)

print("\nLocal minimum point:")
print("x =", round(x, 6))
print("y =", round(function(x), 6))
```
